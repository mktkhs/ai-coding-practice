```python
df.groupby(["日付","営業担当者"]).sum()
```

```python
schedule_table = df.pivot(
    index="日付",
    columns="営業担当者",
    values="予定"
)

schedule_table
```

```python
df_expanded = (
    df.assign(予定=df["予定"].str.split("、")) # 予定列のリストを"、"区切りで分解しリストに格納 assignで変更箇所を置き換えたDataFrameを返す
      .explode("予定")
      .reset_index(drop=True)
)

df_expanded.head()
```

```python
pd.pivot_table(
    df_expanded,
    index="日付",
    columns="営業担当者",
    values="予定",
    aggfunc=lambda schedules: "、".join(schedules),
    # aggfunc=list # リスト型で持つ場合はこっち
    fill_value=""
)
```
