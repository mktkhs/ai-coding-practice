### ラベルを指定してデータを選択

locメソッドでは、DataFrame内で条件を満たした行、列を抽出できる。<br>
通常の参照と異なり、行と列両方の条件を指定できる。

```python
# locメソッドで、行インデックスを指定してデータを列として取得可能
# 例えば、2026年9月1日（index0番目）のデータを取得する場合は、以下のように記述する
df.loc[dates[0]]
```

```python
# 単一データの取得も可能
df.loc[dates[0], "A"]

# 単一データ取得の場合、locよりatメソッドの方が高速に取得可能
df.at[dates[0], "A"]
```

```python
# 列名で指定する場合は、locメソッドの第2引数に列名を渡す
# この時第1引数に:を指定することで、全行を取得することができる
df.loc[:, ["A", "C"]]
# df[["A", "C"]] と出力結果は同じ
```

```python
# 第1引数にindex番号を指定すれば、特定の行のデータを取得可能
df.loc[dates[1]:dates[3], ["A", "C"]]

# df.loc["20260902":"20260904", ["A", "C"]]
# でも同様の結果になる
```

```python
# 条件指定しての抽出も可能
# 行の条件表記なので第1引数に記載
df.loc[df["A"] > 0]
```

```python
# 条件指定で絞り込んだうえで、特定の列を取得することも可能
# 列を指定するなら単純に第2引数を指定すればOK
df.loc[df["A"] > 0, ["C"]]
```

```python
# locを使って行で絞り込む方法はあるのか？ →ある
# 真偽値リストを使ったデータ指定が役に立つ
# 列の条件表記なので第2引数に記載、第1引数は全表示の:を指定する
df.loc[:, df.loc[dates[0]] > 0]
```

```python
# ちなみに列指定の中身はこのようになっている
df.loc[dates[0]] > 0
```

```python
# さらに表示する行を絞る場合はこのように記載
# 行を指定するなら単純に第1引数を指定すればOK
df.loc[[dates[0]], df.loc[dates[0]] > 0]

# 返り値がSeries型でよければ下記でOK
# df.loc[dates[0], df.loc[dates[0]] > 0]
```

## 4. 欠損データ (Missing data)

NumPyにおいて、np.nanは欠損値を表す。
デフォルトでは計算に組み込まれない。

インデックスを再作成すると、指定の軸のインデックスを変更/追加/削除可能。返り値はデータのコピー。

```python
df
```

#### reindexで作り直し

```python
# dfを参照し、5番目までの行名(index)と、すべての列名にGを追加したものを取得
# reindexが通常の参照と異なるのは、行名や列名を指定して新しいDataFrameを作成できる点
df1 = df.reindex(index=dates[0:4], columns=list(df.columns) + ["G"])
df1
```

```python
# 2行目までのG列に1を代入
df1.loc[dates[0]:dates[1], "G"] = 1
df1
```

```python
# 通常の参照だと、呼び出しは可能だが行や列の追加はできない。
df.loc[dates[0:4], list(df.columns)]
```

```python
# indexに追加も可能だが、今回はデータ型が日付だったため、単純に+で追加はできない
# やるとしたらunionを使う必要がある
df0 = df.reindex(
    index= df.index.union([pd.Timestamp("20260907")]), 
    columns= list(df.columns) + ["G"])
df0
```

#### `DataFrame.dropna()` でデータが欠落している行を削除

```python
# dropnaはNaNを含むレコード(行)を削除
# 代入していないので上書きではない
df1.dropna(how="any")
```

#### `DataFrame.fillna()` で欠損データを埋める

```python
# NaNを5に置き換える
df1.fillna(value=5)
```

```python
# df1=で代入しない限り、df1の中身は上書きされない
df1
```

#### `DataFrame.isna()` でNaN値の位置を示すブールマスクを取得

```python
pd.isna(df1)
```
